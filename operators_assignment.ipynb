### You are creating a student result calculator.

A student has:

    Mathematics: 85
    English: 72
    Python: 90
    Total possible marks per subject: 100
    Student name: "Ahmed"
    Required passing marks per subject: 50

### Create variables and:

    Calculate the student's total marks.
    Calculate the average marks.
    Calculate the percentage.
    Check whether the student passed Mathematics.
    Check whether the student passed all three subjects.
    Check whether the name contains the letter "A".
    Check whether the average marks are greater than or equal to 80.
    Store the final percentage and check whether it is truthy.

Print all results.

```python
mathematics_score = 85
english_score = 72
python_score = 90
total_possible_marks_per_subject = 100
student_name = "Ahmed"
required_passing_marks = 50

# 1. Calculate the student's total marks.
total_marks = mathematics_score + english_score + python_score
print(f"Student's Total Marks: {total_marks}")

# 2. Calculate the average marks.
average_marks = total_marks / 3
print(f"Student's Average Marks: {average_marks:.2f}")

# 3. Calculate the percentage.
percentage = (total_marks / (3 * total_possible_marks_per_subject)) * 100
print(f"Student's Percentage: {percentage:.2f}%")

# 4. Check whether the student passed Mathematics.
passed_mathematics = mathematics_score >= required_passing_marks
print(f"Passed Mathematics: {passed_mathematics}")

# 5. Check whether the student passed all three subjects.
passed_all_subjects = (mathematics_score >= required_passing_marks and
                       english_score >= required_passing_marks and
                       python_score >= required_passing_marks)
print(f"Passed All Subjects: {passed_all_subjects}")

# 6. Check whether the name contains the letter "A".
student_name_lower = student_name.lower()
name_contains_A = 'A' in student_name
print(f"Name contains 'A': {name_contains_A}")

# 7. Check whether the average marks are greater than or equal to 80.
average_marks_above_80 = average_marks >= 80
print(f"Average marks greater then 80: {average_marks_above_80}")

# 8. Store the final percentage and check whether it is truthy.
final_percentage_truthy = bool(percentage)
print(f"Final Percentage: {final_percentage_truthy}")
```
